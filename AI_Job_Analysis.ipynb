```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\Muskan\Downloads\sample_ai_jobs.csv")
df.head()
```

```python
df.shape  # Number of rows & columns
df.columns  # Column names
df.info()  # Data types and nulls
df.describe()  # Statistical overview
```

```python
df.isnull().sum()  # Null values count
df.duplicated().sum()  # Duplicates count

# Remove duplicates if any:
df = df.drop_duplicates()
```

```python
df = df.dropna()  # (Remove rows with nulls)
```

```python
df.fillna("Unknown", inplace=True)
```

```python
df['job_title'].value_counts().head(10).plot(kind='barh', color='skyblue')
plt.title("Top 10 AI Job Titles")
plt.xlabel("Number of Postings")
plt.ylabel("Job Title")
plt.show()
```

```python
df['company_name'].value_counts().head(10).plot(kind='bar', color='orange')
plt.title("Top AI Hiring Companies")
plt.xlabel("Company")
plt.ylabel("Job Openings")
plt.xticks(rotation=45)
plt.show()
```

```python
print(df.columns)
```

```python
df.rename(columns={
    'company': 'company_name',
    'experience_required': 'job_type'
#if you're using it that way
}, inplace=True)
```

```python
# Most common job titles
df['job_title'].value_counts()

# Locations with most AI jobs
df['location'].value_counts()

# Most common job types
df['job_type'].value_counts()

# Salary ranges
df['salary_range'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Job type distribution
sns.countplot(data=df, x='job_type')
plt.title('Distribution of Job Types')
plt.xticks(rotation=45)
plt.show()

# Top cities for AI jobs
top_locations = df['location'].value_counts().head(5)
top_locations.plot(kind='bar', title='Top 5 AI Job Locations')
plt.ylabel("Job Count")
plt.show()

# Salary Range Distribution
df['salary_range'].value_counts().plot(kind='barh', title='Salary Ranges')
plt.xlabel("Job Count")
plt.show()
```
